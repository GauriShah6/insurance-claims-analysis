```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={"Policyholder":["A","B","C","D","E","F","G","H","I","J"],"Age":[25,42,35,50,29,46,31,55,38,24],"Number_of_claims":[1,2,0,3,1,2,0,4,1,1],"Claim_amount":[12000,25000,0,45000,8000,30000,0,60000,15000,10000]}
df=pd.DataFrame(data)
df
```

```python
total_claims=df["Number_of_claims"].sum()
total_claim_cost=df["Claim_amount"].sum()
average_claim_amount=df.loc[df["Claim_amount"]>0,"Claim_amount"].mean()
claim_frequency=total_claims/len(df)
print("Total number of claims:",total_claims)
print("Total claim cost: Rs",total_claim_cost)
print("Average amount per policyholder with claims: Rs",round(average_claim_amount,2))
print("Claims per policyholer:",round(claim_frequency,2))
```

```python
plt.figure(figsize=(10,5))
plt.bar(df["Policyholder"],df["Claim_amount"])
plt.title("Insurance Claim Amount by Policyholder")
plt.xlabel("Policyholer")
plt.ylabel("Claim Amount Rs")
plt.show()
```

```python
claim_frequency=df["Number_of_claims"].sum()/len(df)
average_claim_severity=(df["Claim_amount"].sum()/df["Number_of_claims"].sum())
print("Claim frequency:",round(claim_frequency,2))
print("Average claim severity: Rs",round(average_claim_severity,2))
```
